# Funnel Analysis Visualization

```python
import matplotlib.pyplot as plt


stages = ['Views', 'Add to cart', 'Purchase']
values = [1404179, 32272, 9682]
```

```python
plt.figure()
plt.bar(stages, values)
plt.title("User Conversion Funnel")
plt.xlabel("Funnel stage")
plt.ylabel("Number of users")

plt.show()
```

```python
drop_off = [
    0,
    (values[0] - values[1]) / values[0] * 100,
    (values[1] - values[2]) / values[1] * 100
]

plt.figure()
plt.bar(stages,drop_off)
plt.title("Drop off percentage at each stage")
plt.xlabel("Funnel stages")
plt.ylabel("Drop-Off percentage")

plt.show()
```

```python
plt.figure()
plt.plot(stages, values, marker='o')
plt.title("Funnel Progression")
plt.xlabel("Stage")
plt.ylabel("Users Remaining")

plt.show()
```

```python
import pandas as pd

df = pd.read_csv("/Users/swayamware/Documents/user_conversion_funnel/Dataset/events.csv")

df['event_time'] = pd.to_datetime(df['timestamp'], unit='ms')

purchases = df[df['event'] == 'transaction']

daily = purchases.groupby(purchases['event_time'].dt.date).size()

plt.figure()
plt.plot(daily)
plt.title("Daily Purchases Trend")
plt.xlabel("Date")
plt.ylabel("Number of Purchases")

plt.show()
```
